# Subplot in Matplotlib 

```python
A = np.linspace(2,10,7)
A
B = np.linspace(4,10,7)
B
plt.plot(A,B)
plt.title("This is Graph")
plt.xlabel("This is horizontal line")
plt.ylabel("This is vertical line")
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
```

# Object Oriented  Way 

```python
var3 = np.linspace(10,20,7)
var3
var4 = np.linspace(20,40,7)
var4
fig =plt.figure()
axis1= fig.add_axes([0.1,0.1,1,1])
axis1.plot(var3,var4)
axis2= fig.add_axes([0.6,0.1,.3,.3])
axis2.plot(var3,var4)
                   
```
